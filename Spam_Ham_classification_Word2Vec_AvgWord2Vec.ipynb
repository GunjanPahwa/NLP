```python
#Creating a Word2Vec from scratch
import pandas as pd
messages=pd.read_csv('spam.csv',encoding='latin-1',sep=',',names=['label','message'],usecols=[0,1])
```

```python
messages=messages.drop(messages.index[0])
```

```python
messages = messages.iloc[1:].reset_index(drop=True)
```

```python
messages.head()
```

```python
from nltk.stem import WordNetLemmatizer
import nltk
import re
lemmatizer=WordNetLemmatizer()
nltk.download('stopwords')
```

```python

corpus=[]
for i in range(0,len(messages)):
    review=re.sub('[^a-zA-Z]',' ',messages['message'][i])
    review=review.lower()
    review=review.split()
    review=[lemmatizer.lemmatize(word) for word in review ]
    review=' '.join(review)
    corpus.append(review)
```

```python
[[i,j,k] for i,j,k in zip(list(map(len,corpus)),corpus,messages['message']) if i<1]
#This applies len() to every element of corpus.
#i → the length of each corpus item
#j → each item in corpus
#k → corresponding item from messages['message']
#You only keep rows where i < 1
#this means that these rows from x are getting deleted,we need the corresponding rows from y and delete those too
```

```python
from nltk import sent_tokenize
from gensim.utils import simple_preprocess
#Convert a document into a list of lowercase tokens, ignoring tokens that are too short or too long.
```

```python
words = [simple_preprocess(text) for text in corpus]
```

```python
#avgword2vec
import numpy as np
def avg_word2vec(doc):
    tokens = doc.split()  # OR simple_preprocess(doc)
    vectors = [model.wv[word] for word in tokens if word in model.wv.key_to_index]

    if len(vectors) == 0:
        return np.zeros(model.vector_size)   # handle empty sentences

    return np.mean(vectors, axis=0)
```
