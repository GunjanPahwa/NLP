```python
corpus="""Hello Welcome, this is Gunjan trying 100 days of code challenge,
I will try to be consistent on github! Atleast one commit everyday. """
```

```python
print(corpus)
```

```python
#Tokenization
#sentence->paragraphhs
from nltk.tokenize import sent_tokenize #converts paragraphs to sentences
```

```python
documents=sent_tokenize(corpus) #it is considering a new line after exclamation mark and full stop
```

```python
type(documents)
```

```python
for sentence in documents:
    print(sentence)
```

```python
#Tokenization
#paragraph->words
#sentence->words
from nltk.tokenize import word_tokenize
```

```python
word_tokenize(corpus) #every single character including comma, full stop and exclamation mark have been considered as words 
```

```python
#but if i had a word like Gunjan's then it would consider Gunjan and 's as two different words
for sentence in documents:
    print(word_tokenize(sentence))
```

```python
from nltk.tokenize import wordpunct_tokenize
```

```python
wordpunct_tokenize(corpus)
#if i had a word like gunjan's then it would have been splitted as gunjan, ' , s here
```

```python
from nltk.tokenize import TreebankWordTokenizer
```

```python
tokenizer=TreebankWordTokenizer()
# it wont considerfull stops as separate words, only the last full stop would be considered as a separate word.
#this is because after the other full stops we have a new line but the corpus ends after the last full stop
```

```python
tokenizer.tokenize(corpus)
```
