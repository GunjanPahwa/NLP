```python
#classification problem
#whether comments on a product leave a positive review or negative review
#dataset will have reviews ---> eating,eat,eaten (root word is eat)
#but words like this wont affect the final output
#we dont need to have a large number of similar sort of words as it will increase the input features

words=['eating','eats','eaten','writing','writes','programming','programs','history','finally','finalize']
```

```python
##PorterStemmer
from nltk.stem import PorterStemmer
```

```python
stemming=PorterStemmer()
```

```python
for word in words:
    print(word+"---->"+stemming.stem(word))
```

```python
stemming.stem('congratulations')
```

```python
##Regexpstemmer
from nltk.stem import RegexpStemmer
```

```python
reg_stemmer=RegexpStemmer('ing$|s$|e$|able$',min=4)#this means that these suffixes would be removed given the condition that the word length is atleast 2
```

```python
reg_stemmer.stem('eating')
```

```python
reg_stemmer.stem('ingeating')
```

```python
##SnowballStemmer
from nltk.stem import SnowballStemmer
```

```python
snowballsstemmer=SnowballStemmer('english')
```

```python
for word in words:
    print(word+"------->"+snowballsstemmer.stem(word))
```

```python
snowballsstemmer.stem('congratulation')
```

```python
stemming.stem('fairly')
```

```python
snowballsstemmer.stem('fairly')
```
